### Matrix Multiplication

```python
# element wise multiplication
print(tensor, "*", tensor)
print(f"Equals: {tensor*tensor}")
```

```python
# matrix multiplication
print(tensor, "*", tensor)
print(f"Equals: {torch.matmul(tensor, tensor)}")
```

### Matrix Product Vs Element-Wise Product

```python
%%time
value = 0
for i in range(len(tensor)):
    value += tensor[i] * tensor[i]
print(value)
```

```python
%%time
torch.matmul(tensor, tensor)
```

```python
try:
    torch.matmul(torch.rand((3, 2)), torch.rand((3, 2)))
except Exception as e:
    print("Error:", e)
```

### Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - Reshapes an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - Combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - Removes all `1` dimensions from a tensor
* Unsqueeze - Add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

```python
# create a tensor
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(9, 1)
x_reshaped, x_reshaped.shape
```

```python
# change the view
z = x.view(1, 9)
z, z.shape
```

```python
# changing z, changes x, because a view of a tensor shares same memory as the original tensor
z[:, 0] = 5
z, x
```

```python
# stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=1)
x_stacked
```

```python
# squeeze
torch.squeeze(x)
```

## Indexing (Selecting Data From Tensor)

```python
# create a tensor
x = torch.arange(1, 10).reshape(1, 3, 3)
x
```

```python
# indexing on this new tensor
x[0]
```

```python
# index on the middle bracket
x[0][0]
```

```python
# index on last dimension
x[0][0][0]
```

```python
# use ":" to select all of a target dimension
x[:, 0]
```

```python
# get all values of 0th and 1st dimensions but only index 1 of 2nd dimension
x[:, :, 1]
```

## PyTorch Tensors and Numpy

```python
# numpy array to tensor
array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array) # creating tensor from existing np array, inherits the dtype of the array
array, tensor
```

```python
tensor.dtype
```

```python
# change value of array, what will it do to the "tensor"?
array = array + 1
array, tensor
```

```python
# Tensor to Numpy array
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
print("Tensor:", tensor)
print("Numpy Tensor:", numpy_tensor)
```
